```scala
val path =  "dbfs:/mnt/dados/inbound/dados_brutos_imoveis.json"
val dados = spark.read.json(path)
```

```scala
val dados_anuncio = dados.drop("imagens","usuario")
```

```scala
display(dados_anuncio)
```

```scala
import org.apache.spark.sql.functions.col 
```

```scala
val df_bronze = dados_anuncio.withColumn("id",col("anuncio.id"))
```

```scala
display(df_bronze)
```

```scala
val path_bronze = "dbfs:/mnt/dados/bronze/dataset_imoveis"
df_bronze.write.format("delta").mode(SaveMode.Overwrite).save(path_bronze)
```

----------------------------------------------------------------------SILVER--------------------------------------------------------------------------------

```scala
val df = spark.read.format("delta").load(path_bronze)
```

```scala
display(df.select("anuncio.*","anuncio.endereco.*").limit(3))
```

```scala
val dados_detalhados = df.select("anuncio.*","anuncio.endereco.*").drop("caracteristicas","endereco")
dados_detalhados.write.format("delta").mode("overwrite").save("dbfs:/mnt/dados/silver/dataset_imoveis")
```
